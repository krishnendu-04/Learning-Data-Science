```python
#function without argument and no return type
def prime1():
    n = int(input("Enter the number: "))
    if n<2:
        print("Not prime")
    else:
        flag = 0
        for i in range(2,n):
            if n%i==0:
                flag = 1
                break
        if flag>0:
            print("Not prime")
        else:
            print("Prime")

prime1()
```

```python
#function with argument and no return type
def prime2(num):
    if num<2:
            print("Not prime")
    else:
        flag = 0
        for i in range(2,num):
            if num%i==0:
                flag = 1
                break
        if flag>0:
            print("Not prime")
        else:
            print("Prime")
prime2(int(input("Enter the number: ")))
```

```python
#function with argument and return type
def prime3(num):
    if num<2:
            print("Not prime")
    else:
        flag = 0
        for i in range(2,num):
            if num%i==0:
                flag = 1
                break
    return flag
flag = prime3(int(input("Enter the number: ")))
if flag>0:
    print("Not prime")
else:
    print("Prime")
```

```python
#function without argument and no return type
def sum1():
    lower = int(input("Enter the lower limit: "))
    upper = int(input("Enter the upper limit: "))
    even_sum = 0
    odd_sum = 0
    for i in range(lower,upper+1):
        if i%2==0:
            even_sum+=i
        else:
            odd_sum+=i
    print("Even Sum: ",even_sum,"\nOdd Sum: ",odd_sum)
sum1()
```

```python
#function with argument and no return type
def sum2(lower,upper):
    even_sum = 0
    odd_sum = 0
    for i in range(lower,upper+1):
        if i%2==0:
            even_sum+=i
        else:
            odd_sum+=i
    print("Even Sum: ",even_sum,"\nOdd Sum: ",odd_sum)
sum2(int(input("Enter the lower limit: ")),int(input("Enter the upper limit: ")))
```

```python
#function with argument and return type
def sum3(lower,upper):
    even_sum = 0
    odd_sum = 0
    for i in range(lower,upper+1):
        if i%2==0:
            even_sum+=i
        else:
            odd_sum+=i
    return even_sum,odd_sum
even,odd = sum3(int(input("Enter the lower limit: ")),int(input("Enter the upper limit: ")))
print("Even Sum: ",even,"\nOdd Sum: ",odd)
```

```python
def add(n1,n2):
    return n1+n2

def sub(n1,n2):
    return n1-n2

def mul(n1,n2):
    return n1*n2

def div(n1,n2):
    return n1/n2

n1 = int(input("Enter the number: "))
n2 = int(input("Enter the number: "))
choice = int(input("1. Addition\n2.  Subtraction\n3. Multiplication\n4. Division\nEnter your choice: "))
if choice==1:
    print(add(n1,n2))
elif choice==2:
    print(sub(n1,n2))
elif choice==3:
   print(mul(n1,n2))
elif choice==4:
    print(div(n1,n2))
else:
    print("Invalid choice")
```

```python
def calculator(choice,n1,n2):
    if choice==1:
        return n1+n2
    elif choice==2:
        return n1-n2
    elif choice==3:
        return n1*n2
    elif choice==4:
        if n2==0:
            print("Division by zero not possible")
        return n1/n2
    else:
        print("Invalid choice")   

print("1. Addition\n2. Subtraction\n3. Multiplication\n4. Division")
n1 = int(input("Enter the number: "))
n2 = int(input("Enter the number: "))
choice = int(input("\nEnter your choice: "))
data = calculator(choice,n1,n2)
print("Result: ",data)
```
