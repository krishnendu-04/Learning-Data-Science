```python
def prime1():
    n = int(input("Enter the number: "))
    if n<2:
        print("Not prime")
    else:
        flag = 0
        for i in range(2,n):
            if n%i==0:
                flag = 1
                break
        if flag>0:
            print("Not prime")
        else:
            print("Prime")

prime1()
```

```python
def prime2(num):
    if num<2:
            print("Not prime")
    else:
        flag = 0
        for i in range(2,num):
            if num%i==0:
                flag = 1
                break
        if flag>0:
            print("Not prime")
        else:
            print("Prime")
prime2(int(input("Enter the number: ")))
```

```python
def fact1():
    n = int(input("Enter the number: "))
    fact=1
    if n<=0:
        print(0)
    else:
        for i in range(1,n+1):
            fact*=i
        print(fact)
fact1()
```
