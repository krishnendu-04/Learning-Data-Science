```python
n = int(input("Enter the number: "))
num = n
rev = 0
while(n>0):
    a = n%10
    rev = rev*10+a
    n//=10
if rev==num:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
n = int(input("Enter the number: "))
temp = n
sum = 0
while(n>0):
    a = n%10
    sum += a**3
    n//=10
if temp==sum:
    print("Armstrong")
else:
    print("Not armstrong")
```

```python
n = int(input("Enter the number: "))
sum = 0
for i in range(1,n):
    if n%i==0:
        sum+=i
if sum==n:
    print("Perfect Number")
else:
    print("Imperfect number")
```

```python
n = int(input("Enter the number: "))
a = n
sum = 0
while(n>0):
    last_digit = n%10
    fact = 1
    for i in range(1,last_digit+1):
        fact*=i
    sum+=fact
    n//=10
if a==sum:
    print("Strong Number")
else:
    print("Not a strong number")
```

```python
n = int(input("Enter the number: "))
num = n
sq = n*n
count = 0

while(n>0):
    count+=1
    n//=10

pow = 10 ** count

if sq%pow == num:
    print("Automorphic Number")
else:
    print("Not an Automorphic Number")
```

```python
n = int(input("Enter the number: "))
sq = n*n
sum = 0
while sq>0:
    digit = sq%10
    sum+=digit
    sq//=10
if sum==n:
    print("Neon Number")
else:
    print("Not Neon Number")
```

```python
n = int(input("Enter the number: "))
sum = 0
prod = 1
while(n>0):
    digit = n%10
    sum+=digit
    prod*=digit
    n//=10
if sum==prod:
    print("Spy Number")
else:
    print("Not a Spy Number")
```
