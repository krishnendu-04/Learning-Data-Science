```python
import pandas as pd
```

```python
name = ['anu','angana','fidha']
s1 = pd.Series(name)
print(s1)
```

```python
s2 = pd.Series([10,20,30,40],index=['a','b','c','d'])
print(s2)
```

```python
s3 = pd.Series(['Vishnu',30,'Thrissur'],index=['Name','Age','Place'])
print(s3)
print(s3['Name'])
print(s3.iloc[0])
print(s3.iloc[1])
print(s3.iloc[2])
print(s3.iloc[0:])
```

```python
#s4 = pd.Series([i for i in range(1,101,1)])
import numpy as np
s4 = pd.Series(np.arange(1,101))
print(s4)
```

```python
lst1 = [['Anu',23,'Ernakulam'],['Fidha',22,'Ernakulam'],['Parvathy',22,'Thrissur']]
df1 = pd.DataFrame(lst1)
print(df1)
print(df1.shape)
print(df1.ndim)
print(df1.dtypes)
```

```python
dict1 = [{"name":"Anu","age":23,"place":"Ernakulam"},{"name":"Fidha","age":22,"place":"Ernakulam"},{"name":"Parvathy","age":22,"place":"Thrissur"}]
df2 = pd.DataFrame(dict1)
print(df2)
print(df2.shape)
print(df2.ndim)
print(df2.dtypes)
```

```python
dict1 = {'name':['Anu','Fidha','Parvathy'],
        'age':[23,22,22],
        'place':['Ernakulam','Ernakulam','Thrissur']}
df3 = pd.DataFrame(dict1)
print(df3)
print(df3.shape)
print(df3.ndim)
print(df3.dtypes)
```
